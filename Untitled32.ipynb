```python

import numpy as np

# 1. Objective function: Traffic delay
def fitness(x):
    return (x - 40)**2 + 10


# 2. Initialize parameters
rows = 5
cols = 5
Max_iter = 10

lower_bound = 10
upper_bound = 90

np.random.seed(42)


# 3. Initialize cells with random green times
grid = np.random.uniform(
    lower_bound, upper_bound, (rows, cols)
)

# 4. Calculate initial fitness
scores = fitness(grid)

best_index = np.argmin(scores)
best_solution = grid.flat[best_index]
best_fitness = scores.flat[best_index]


# 5. Main PCA algorithm
for iteration in range(Max_iter):

    new_grid = grid.copy()

    for i in range(rows):
        for j in range(cols):

            neighbours = []

            # Find neighbouring cells
            for x in range(max(0, i-1), min(rows, i+2)):
                for y in range(max(0, j-1), min(cols, j+2)):

                    if x != i or y != j:
                        neighbours.append(grid[x, y])

            # Find average neighbour value
            average = np.mean(neighbours)

            # Update cell state
            new_grid[i, j] = average

    # Update the complete grid
    grid = np.clip(new_grid, lower_bound, upper_bound)

    # 6. Evaluate fitness
    scores = fitness(grid)

    # Find best cell
    best_index = np.argmin(scores)

    current_best = grid.flat[best_index]
    current_fitness = scores.flat[best_index]

    # Update global best
    if current_fitness < best_fitness:
        best_solution = current_best
        best_fitness = current_fitness

    print(
        "Iteration:", iteration + 1,
        "Green Time:", round(best_solution, 2),
        "Traffic Delay:", round(best_fitness, 2)
    )


# 7. Final output
print("\nFINAL RESULT")
print("Optimal Green Signal Time:",
      round(best_solution, 2), "seconds")

print("Minimum Traffic Delay:",
      round(best_fitness, 2))
```

```python

import numpy as np
import math

weights = np.array([5, 4, 3, 6, 2])
values = np.array([30, 25, 15, 20, 10])

capacity = 15
n_items = len(weights)


def fitness(solution):
    total_weight = np.sum(solution * weights)
    total_value = np.sum(solution * values)

    if total_weight > capacity:
        return total_value - 1000 * (total_weight - capacity)

    return total_value


def levy_flight():
    beta = 1.5

    sigma = (
        math.gamma(1 + beta) * np.sin(np.pi * beta / 2)
        / (
            math.gamma((1 + beta) / 2)
            * beta * 2 ** ((beta - 1) / 2)
        )
    ) ** (1 / beta)

    u = np.random.normal(0, sigma)
    v = np.random.normal(0, 1)

    return u / (abs(v) ** (1 / beta) + 1e-10)


def convert_to_binary(position):
    probabilities = 1 / (1 + np.exp(-np.clip(position, -50, 50)))

    solution = (np.random.rand(n_items) < probabilities).astype(int)

    return solution


n = 5
Pa = 0.25
Max_iter = 50

np.random.seed(42)


nests = np.random.randint(0, 2, size=(n, n_items))

scores = np.array([fitness(x) for x in nests])

best_index = np.argmax(scores)

best_solution = nests[best_index].copy()
best_fitness = scores[best_index]


for iteration in range(Max_iter):

    for i in range(n):

        position = nests[i] + 0.1 * levy_flight() * (
            nests[i] - best_solution
        )

        new_solution = convert_to_binary(position)

        new_fitness = fitness(new_solution)

        j = np.random.randint(0, n)

        if new_fitness > scores[j]:
            nests[j] = new_solution.copy()
            scores[j] = new_fitness

    number_to_abandon = max(1, int(Pa * n))

    worst_indices = np.argsort(scores)[:number_to_abandon]

    for j in worst_indices:
        nests[j] = np.random.randint(0, 2, n_items)
        scores[j] = fitness(nests[j])

    best_index = np.argmax(scores)

    if scores[best_index] > best_fitness:
        best_solution = nests[best_index].copy()
        best_fitness = scores[best_index]

    print(
        "Iteration:", iteration + 1,
        "Best Solution:", best_solution,
        "Best Value:", best_fitness
    )


total_weight = np.sum(best_solution * weights)
total_value = np.sum(best_solution * values)

print("\nFINAL RESULT")
print("Selected Items:", best_solution)
print("Total Weight:", total_weight, "kg")
print("Total Value:", total_value)
```

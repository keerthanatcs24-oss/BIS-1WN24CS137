```python

import random

dist = [
    [0, 2, 9, 10],
    [2, 0, 6, 4],
    [9, 6, 0, 3],
    [10, 4, 3, 0]
]

random.seed(42)

n = len(dist)
ants = 4
iterations = 5
alpha = 1
beta = 2
rho = 0.5

pheromone = [
    [0 if i == j else 1.0 for j in range(n)]
    for i in range(n)
]

best_tour = None
best_distance = float("inf")

for iteration in range(iterations):
    print("\nIteration", iteration + 1)
    tours = []

    for ant in range(ants):
        start = random.randrange(n)
        tour = [start]
        unvisited = set(range(n))
        unvisited.remove(start)

        while unvisited:
            current = tour[-1]
            choices = list(unvisited)

            weights = [
                pheromone[current][c] ** alpha
                * (1 / dist[current][c]) ** beta
                for c in choices
            ]

            next_city = random.choices(
                choices, weights=weights, k=1
            )[0]

            tour.append(next_city)
            unvisited.remove(next_city)

        tour.append(start)

        distance = sum(
            dist[tour[i]][tour[i + 1]]
            for i in range(len(tour) - 1)
        )

        tours.append((tour, distance))

        print(
            f"Ant {ant + 1} Tour: {tour} "
            f"Distance: {distance}"
        )

        if distance < best_distance:
            best_distance = distance
            best_tour = tour[:]

    # Evaporate pheromones
    for i in range(n):
        for j in range(n):
            if i != j:
                pheromone[i][j] *= (1 - rho)

    # Deposit pheromones
    for tour, distance in tours:
        deposit = 1 / distance

        for i in range(len(tour) - 1):
            a = tour[i]
            b = tour[i + 1]
            pheromone[a][b] += deposit
            pheromone[b][a] += deposit

print("\nBest Tour:", best_tour)
print("Shortest Distance:", best_distance)
```

```python
import random

# Parameters
POP_SIZE = 6
GENES = 5
MUTATION_RATE = 0.1
GENERATIONS = 5

# Fitness function: maximize x²
def fitness(chromosome):
    x = int("".join(map(str, chromosome)), 2)
    return x * x


# Create random population
population = [
    [random.randint(0, 1) for _ in range(GENES)]
    for _ in range(POP_SIZE)
]

print("Initial Population:")
for p in population:
    print(p)

# Genetic Algorithm
for generation in range(GENERATIONS):

    print("\nGeneration", generation + 1)

    # Evaluate fitness
    population.sort(key=fitness, reverse=True)

    print("Best:", population[0],
          "Fitness:", fitness(population[0]))

    # Selection
    parents = population[:POP_SIZE // 2]

    # Crossover
    children = []

    while len(children) < POP_SIZE:

        p1 = random.choice(parents)
        p2 = random.choice(parents)

        point = random.randint(1, GENES - 1)

        child = p1[:point] + p2[point:]

        print("Crossover:")
        print("Parent 1:", p1)
        print("Parent 2:", p2)
        print("Child   :", child)

        # Mutation
        for i in range(GENES):
            if random.random() < MUTATION_RATE:
                child[i] = 1 - child[i]

        children.append(child)

    population = children


# Final best solution
best = max(population, key=fitness)

print("\nFinal Best Solution:")
print("Chromosome:", best)
print("Fitness:", fitness(best))
print("Value:", int("".join(map(str, best)), 2))
```

```python
import random

# Parameters
population_size = 6
mutation_rate = 0.1
generations = 3

# Light, Fan, TV, AC
power = [1, 2, 2, 5]


# Create random population
def generate_population():
    return [
        [random.randint(0, 1) for _ in range(4)]
        for _ in range(population_size)
    ]


# Calculate electricity consumption
def consumption(chromosome):
    return sum(chromosome[i] * power[i] for i in range(4))


# Fitness: lower consumption = better
def fitness(chromosome):
    return 1 / (1 + consumption(chromosome))


# Crossover
def crossover(p1, p2):
    point = random.randint(1, 3)

    child = p1[:point] + p2[point:]

    print("  Parent 1:", p1)
    print("  Parent 2:", p2)
    print("  Crossover point:", point)
    print("  Child:", child)

    return child


# Mutation
def mutate(child):
    for i in range(4):
        if random.random() < mutation_rate:
            child[i] = 1 - child[i]

    return child


# ---------------- MAIN ----------------

population = generate_population()

print("INITIAL RANDOM POPULATION")
print(population)


for generation in range(1, generations + 1):

    print("\n========== GENERATION", generation, "==========")

    # Fitness
    print("\nFitness:")
    for chromosome in population:
        print(
            chromosome,
            "Consumption =", consumption(chromosome),
            "Fitness =", round(fitness(chromosome), 3)
        )

    # Selection
    population.sort(key=fitness, reverse=True)
    parents = population[:population_size // 2]

    print("\nSELECTED PARENTS:")
    for p in parents:
        print(p)

    # Crossover + Mutation
    children = []

    print("\nCROSSOVER:")

    while len(children) < population_size:

        p1 = random.choice(parents)
        p2 = random.choice(parents)

        child = crossover(p1, p2)

        before = child.copy()
        child = mutate(child)

        print("  Before Mutation:", before)
        print("  After Mutation :", child)

        children.append(child)

    population = children


# Best solution
best = min(population, key=consumption)

print("\n========== FINAL RESULT ==========")
print("Best Schedule:", best)
print("Minimum Consumption:", consumption(best), "units")

print("\nAppliances:")
print("Light:", "ON" if best[0] else "OFF")
print("Fan  :", "ON" if best[1] else "OFF")
print("TV   :", "ON" if best[2] else "OFF")
print("AC   :", "ON" if best[3] else "OFF")
```

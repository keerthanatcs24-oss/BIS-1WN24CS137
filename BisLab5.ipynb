```python

import numpy as np

def fitness(x):
    return np.sum(x ** 2)

n_wolves = 5
dimensions = 2
max_iterations = 10

lower_bound = -10
upper_bound = 10

wolves = np.random.uniform(
    lower_bound, upper_bound,
    (n_wolves, dimensions)
)

alpha_pos = np.zeros(dimensions)
beta_pos = np.zeros(dimensions)
delta_pos = np.zeros(dimensions)

alpha_score = float("inf")
beta_score = float("inf")
delta_score = float("inf")

for iteration in range(max_iterations):

    for i in range(n_wolves):

        score = fitness(wolves[i])

        if score < alpha_score:
            delta_score = beta_score
            delta_pos = beta_pos.copy()

            beta_score = alpha_score
            beta_pos = alpha_pos.copy()

            alpha_score = score
            alpha_pos = wolves[i].copy()

        elif score < beta_score:
            delta_score = beta_score
            delta_pos = beta_pos.copy()

            beta_score = score
            beta_pos = wolves[i].copy()

        elif score < delta_score:
            delta_score = score
            delta_pos = wolves[i].copy()

    a = 2 - iteration * (2 / max_iterations)

    for i in range(n_wolves):

        for d in range(dimensions):

            r1 = np.random.rand()
            r2 = np.random.rand()

            A1 = 2 * a * r1 - a
            C1 = 2 * r2

            D_alpha = abs(C1 * alpha_pos[d] - wolves[i][d])
            X1 = alpha_pos[d] - A1 * D_alpha

            r1 = np.random.rand()
            r2 = np.random.rand()

            A2 = 2 * a * r1 - a
            C2 = 2 * r2

            D_beta = abs(C2 * beta_pos[d] - wolves[i][d])
            X2 = beta_pos[d] - A2 * D_beta

            r1 = np.random.rand()
            r2 = np.random.rand()

            A3 = 2 * a * r1 - a
            C3 = 2 * r2

            D_delta = abs(C3 * delta_pos[d] - wolves[i][d])
            X3 = delta_pos[d] - A3 * D_delta

            wolves[i][d] = (X1 + X2 + X3) / 3

        wolves[i] = np.clip(
            wolves[i], lower_bound, upper_bound
        )

    print("Iteration:", iteration + 1,
          "Best Fitness:", round(alpha_score, 6))

print("\nBest Solution (Alpha):", alpha_pos)
print("Minimum Fitness:", alpha_score)
```
